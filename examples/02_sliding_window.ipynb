# Sliding-window feature extraction

Compute features for overlapping windows of a signal.

```python
import numpy as np

from tscv_vision.sliding import features_for_sliding

t = np.linspace(0, 8*np.pi, 256)
x = np.sin(t)
feats, starts = features_for_sliding(x, encoder="gaf", size=64, hop=32)
feats.shape, starts
```
