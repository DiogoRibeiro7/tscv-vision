# Interpretability demo

```python
import numpy as np
from tscv_vision.analytics import TSHAPExplainer, gaf_attribution, plot_importance, generate_counterfactual
model = lambda x: float(x.mean())
series = np.sin(np.linspace(0, 2*np.pi, 64))
```

```python
explainer = TSHAPExplainer(model)
t_imp, f_imp = explainer.explain(series, window=16)
t_imp[:5]
```

```python
fig = plot_importance(series, t_imp)
fig.show()
```

```python
imp_mat = np.ones((64, 64))
gaf_attribution(imp_mat)[:5]
```

```python
cf = generate_counterfactual(series, model, target=0.5)
cf[:5]
```
